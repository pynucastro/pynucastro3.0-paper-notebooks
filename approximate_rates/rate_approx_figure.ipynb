## C/O burning

```python
nuclei = ["p", "he4", "o16", "si28", "p31", "s32"]
net = pyna.network_helper(nuclei)
```

```python
o16 = pyna.Nucleus("o16")
si28 = pyna.Nucleus("si28")
s32 = pyna.Nucleus("s32")
p31 = pyna.Nucleus("p31")
```

```python
nuclei_custom_labels = {o16: "A", si28: "B", s32: "C", p31: "X"}
edge_labels = {(o16, p31): r"$\lambda_{A(A,p)X}$",
               (o16, si28): r"$\lambda_{A(A,\alpha)B}$",
               (p31, si28): r"$\lambda_{X(p,\alpha)B}$",
               (p31, s32): r"$\lambda_{X(p,\gamma)C}$",
               (si28, s32): r"$\lambda_{B(\alpha,\gamma)C}$"}
```

```python
fig = net.plot(rotated=True, curved_edges=True,
               edge_labels=edge_labels, curved_edge_radius=0.14, edge_labels_size=16,
               nuclei_custom_labels=nuclei_custom_labels,
               node_size=1200, node_font_size=20, size=(900, 550))
fig.tight_layout()
ax = fig.gca()
ax.get_xaxis().set_ticks([])
ax.get_yaxis().set_ticks([])
ax.set_ylim([-0.2, 1.1])
fig.savefig("approx4.png", bbox_inches="tight")
```
